```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_parquet('../data/processed/clean_soil_data.parquet')
print(f"Clean records: {len(df)}")
df.describe().T
```

## 1. Target Distributions & Skewness

```python
targets = ['lead_mg_kg', 'cadmium_mg_kg', 'chromium_mg_kg', 'nickel_mg_kg']
df[targets].skew()
```

## 2. Soil Properties vs Heavy Metal Interactions

```python
plt.figure(figsize=(10, 8))
sns.scatterplot(data=df, x='soil_ph', y='chromium_mg_kg', hue='land_use', size='clay_pct')
plt.title('Soil pH vs Chromium (Cr) by Land Use')
plt.show()
```
